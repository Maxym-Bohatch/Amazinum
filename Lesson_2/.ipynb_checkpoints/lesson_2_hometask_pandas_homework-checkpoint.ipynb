```python
import pandas as pd
import numpy as np

def answer_one():
    energy = pd.read_excel('Energy Indicators.xls', skiprows=17, skipfooter=38)
    energy = energy.iloc[:, [2, 3, 4, 5]]
    energy.columns = ['Country', 'Energy Supply', 'Energy Supply per Capita', '% Renewable']
    
    energy = energy.replace('...', np.nan)
    energy['Energy Supply'] = pd.to_numeric(energy['Energy Supply']) * 1000000
    energy['Energy Supply per Capita'] = pd.to_numeric(energy['Energy Supply per Capita'])

    energy['Country'] = energy['Country'].str.replace(r'\d+', '', regex=True)
    energy['Country'] = energy['Country'].str.replace(r'\s*\(.*\)', '', regex=True).str.strip()
    
    energy['Country'] = energy['Country'].replace({
        "Republic of Korea": "South Korea",
        "United States of America": "United States",
        "United Kingdom of Great Britain and Northern Ireland": "United Kingdom",
        "China, Hong Kong Special Administrative Region": "Hong Kong"
    })

    gdp = pd.read_excel('world_bank.xls', skiprows=3)
    
    gdp['Country Name'] = gdp['Country Name'].replace({
        "Korea, Rep.": "South Korea", 
        "Iran, Islamic Rep.": "Iran",
        "Hong Kong SAR, China": "Hong Kong"
    })
    
    years = [str(y) for y in range(2006, 2016)]
    gdp = gdp[['Country Name'] + years]
    gdp.columns = ['Country'] + years

    scimen = pd.read_excel('scimagojr.xlsx')
    scimen_top15 = scimen.iloc[:15]

    df = pd.merge(scimen_top15, energy, how='inner', on='Country')
    final_df = pd.merge(df, gdp, how='inner', on='Country')
    final_df = final_df.set_index('Country')
    
    return final_df

Top15 = answer_one()
Top15.head()
```

```python
def answer_two():
    Top15 = answer_one()
    years = [str(y) for y in range (2006, 2016)]
    avgGDP = Top15[years].mean(axis=1)
    return avgGDP.sort_values(ascending=False)
avgGDP = answer_two()
avgGDP
```

```python
def answer_three():
    Top15=answer_one()
    target_country = answer_two().index[5]
    country_data = Top15.loc[target_country]
    return country_data['2015'] - country_data['2006']
diff = answer_three()
diff
```

```python
def answer_four():
    Top15 = answer_one()
    return Top15['Energy Supply per Capita'].mean()
```

```python
def answer_five():
    Top15 = answer_one()
    country = Top15['% Renewable'].idxmax()
    value = Top15['% Renewable'].max()
    return (country, value)
```

```python
def answer_six():
    Top15 = answer_one()
    median_val = Top15['% Renewable'].median()
    Top15['HighRenew'] = Top15['% Renewable'].apply(lambda x:1 if x>=median_val else 0)
    return Top15['HighRenew'].sort_index(ascending=True)
```

```python
def answer_seven():
    Top15 = answer_one()
    ContinentDict ={'China':'Asia', 'United States':'North America', 'Japan':'Asia', 
                      'United Kingdom':'Europe', 'Russian Federation':'Europe', 
                      'Canada':'North America', 'Germany':'Europe', 'India':'Asia',
                      'France':'Europe', 'South Korea':'Asia', 'Italy':'Europe', 
                      'Spain':'Europe', 'Iran':'Asia', 'Australia':'Australia', 
                      'Brazil':'South America'}
    Top15['PopEst'] = Top15['Energy Supply'] / Top15['Energy Supply per Capita']
    Top15['Continent'] = Top15.index.to_series().map(ContinentDict)
    result = Top15.groupby('Continent')['PopEst'].agg(['size', 'sum', 'mean', 'std'])
    return result
```
