```python
#Імовірність результату «H» («Голова») під час підкидання кожної з 4 монет
#(назвемо їх c1, c2, c3, c4) зі зміненим центром ваги дорівнює [0,12, 0,27, 0,21, 0,96] відповідно. 
#Навмання вибрали одну з монет, і почалися випробування. 
#Усі подальші тести виконуються для тієї ж монети. 
#Визначте ймовірність «H» у наступному підкиданні після 9 фактично виконаних тестів: [H H H T H T H H H] 
#(тут «T» («хвіст») — протилежна сторона медалі). 
#Наприклад, перед першим тестом ймовірність «Н» становить 0,39 
#(за формулою повної ймовірності з урахуванням еквівалентності вибору однієї з доступних монет). 
#Маючи докази «H» у першому тесті, ймовірність гіпотези, що вибрана монета є c1 / c2 / c3, зменшилася, 
#а ймовірність гіпотези, що c4 збільшилася, і, отже, ймовірність підкинути «H» у наступному тесті тепер дорівнює 0,67. 
#Подібним чином, після випуску «H» в іншому тесті, вам потрібно повторно обчислити ймовірність перевернути «H» під час третього підкидання і так далі. 
#Згадайте правило Байєса. Формат відповіді:
#список ймовірностей з точністю до сотих [0,67,?,? ,?,? ,?,? ,? ] (замініть «?» на відповідні значення) ● Код програми або опис
```

```python
import numpy as np
coins_h = np.array([0.12, 0.27, 0.21, 0.96]) #фіксовані шанси монет
results = ["H", "H", "H", "T", "H", "T", "H", "H", "H"] #послідовність того,що випало
priors = np.array([0.25, 0.25, 0.25, 0.25]) #початкова віра в кожну монету
final_list = [] # список результатів
for res in results:# визначення правди
    if res == "H":
        likelihoods = coins_h
    else:
        likelihoods = 1 - coins_h
    priors = priors * likelihoods #множення попередніх знаннь на нові докази ймовірності
    priors = priors / np.sum(priors) #ділення на суму для отримання імоірності
    next_h = np.sum(priors * coins_h) #тут робиться прогноз на наступний кидок (шанс кожної монети*на віру в те що це вона)
    final_list.append(round(float(next_h), 2))#піся всього додаємо в список перетворюючи на звичайне число з і значенням після крапки
print("Results:", final_list)
```
